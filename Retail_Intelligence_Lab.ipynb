```python
query = """
SELECT
    COUNT(*) AS total_rows,
    SUM(CASE
        WHEN InvoiceNo LIKE 'C%' THEN 1
        ELSE 0
    END) AS cancellation_rows,
    SUM(CASE
        WHEN Quantity < 0 THEN 1
        ELSE 0
    END) AS negative_quantity_rows,
    SUM(CASE
        WHEN CustomerID IS NULL THEN 1
        ELSE 0
    END) AS rows_without_customer_id,
    SUM(CASE
        WHEN UnitPrice <= 0 THEN 1
        ELSE 0
    END) AS zero_or_negative_price_rows
FROM retail;
"""

quality_result = pd.read_sql_query(query, conn)
display(quality_result)
```

```python

query = """
SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_units_purchased,
    ROUND(SUM(Quantity * UnitPrice), 2) AS total_spend,
    ROUND(
        SUM(Quantity * UnitPrice) /
        COUNT(DISTINCT InvoiceNo),
        2
    ) AS average_order_value
FROM retail
WHERE InvoiceNo NOT LIKE 'C%'
  AND Quantity > 0
  AND UnitPrice > 0
  AND CustomerID IS NOT NULL
GROUP BY CustomerID
HAVING SUM(Quantity * UnitPrice) > 0
ORDER BY total_spend DESC
LIMIT 10;
"""

top_customers = pd.read_sql_query(query, conn)
display(top_customers)
```

```python

query = """
WITH customer_orders AS (
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS total_orders,
        ROUND(SUM(Quantity * UnitPrice), 2) AS total_spend
    FROM retail
    WHERE InvoiceNo NOT LIKE 'C%'
      AND Quantity > 0
      AND UnitPrice > 0
      AND CustomerID IS NOT NULL
    GROUP BY CustomerID
),
customer_segments AS (
    SELECT
        CustomerID,
        total_orders,
        total_spend,
        CASE
            WHEN total_orders = 1 THEN 'One-time'
            WHEN total_orders BETWEEN 2 AND 5 THEN 'Occasional'
            ELSE 'Frequent'
        END AS customer_segment
    FROM customer_orders
)
SELECT
    customer_segment,
    COUNT(*) AS customer_count,
    ROUND(AVG(total_spend), 2) AS average_customer_spend,
    ROUND(SUM(total_spend), 2) AS segment_sales
FROM customer_segments
GROUP BY customer_segment
ORDER BY
    CASE customer_segment
        WHEN 'Frequent' THEN 1
        WHEN 'Occasional' THEN 2
        WHEN 'One-time' THEN 3
    END;
"""

customer_segments = pd.read_sql_query(query, conn)
display(customer_segments)
```

```python

import matplotlib.pyplot as plt

# Sort segments by customer count
segment_chart = customer_segments.sort_values(
    "customer_count", ascending=True
).copy()

# Calculate percentage of customers
total_customers = segment_chart["customer_count"].sum()
segment_chart["customer_pct"] = (
    segment_chart["customer_count"] / total_customers * 100
)

# Premium dark theme
fig, ax = plt.subplots(figsize=(11, 6))
fig.patch.set_facecolor("#101827")
ax.set_facecolor("#101827")

colors = []
for segment in segment_chart["customer_segment"]:
    if segment == "Frequent":
        colors.append("#F59E0B")
    elif segment == "Occasional":
        colors.append("#2DD4BF")
    else:
        colors.append("#64748B")

bars = ax.barh(
    segment_chart["customer_segment"],
    segment_chart["customer_count"],
    color=colors,
    height=0.58
)

# Add count and percentage labels
for bar, (_, row) in zip(bars, segment_chart.iterrows()):
    value = int(row["customer_count"])
    pct = row["customer_pct"]

    ax.annotate(
        f" {value:,} customers  |  {pct:.1f}%",
        xy=(bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(5, 0),
        textcoords="offset points",
        va="center",
        color="#F8FAFC",
        fontsize=10,
        fontweight="bold"
    )

ax.set_title(
    "Customer Segmentation by Order Frequency",
    loc="left",
    color="#F8FAFC",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_xlabel("Number of customers", color="#CBD5E1", labelpad=12)
ax.set_ylabel("")
ax.tick_params(axis="x", colors="#CBD5E1")
ax.tick_params(axis="y", colors="#F8FAFC", labelsize=11)
ax.grid(axis="x", color="#334155", linestyle="--", alpha=0.5)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```
